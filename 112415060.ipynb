Q1.Numpy array operations


```python
import numpy as np

A = np.array([
    [12, 5, 8, 20],
    [7, 15, 3, 11],
    [25, 9, 18, 6],
    [10, 14, 2, 16]
])
mean = np.mean(A, axis=1)
binary_matrix = (A > mean[:, np.newaxis]).astype(int)
count = np.sum(A > mean[:, np.newaxis], axis=1)

max_count = np.max(count)
rows = np.where(count == max_count)[0]

print("Row means:")
print(mean)

print("\nBinary matrix:")
print(binary_matrix)

print("\nCount:")
print(count)

print("\nMaximum count:", max_count)
print("Rows:", rows + 1)
```

Q2.   X = np.array([ [10, 100, 5], [12, 110, 6], [11, 105, 7], [13, 500, 6], [14, 108, 8], [15, 115, 9] ]) For each column: 1. Calculate the mean and standard deviation. 2. Identify values satisfying: ∣x−μ∣>2σ 3. Replace such values with the corresponding column median. 4. Do this without modifying non-outlier values.

```python
import numpy as np

X = np.array([
    [10, 100, 5],
    [12, 110, 6],
    [11, 105, 7],
    [13, 500, 6],
    [14, 108, 8],
    [15, 115, 9]
])
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
outlier_mask = np.abs(X - mean) > 2 * std
median = np.median(X, axis=0)
X_clean = np.where(outlier_mask, median, X)

print(mean)
print(std)
print(outlier_mask)
print(median)
print(X_clean)
```

Q3.  import pandas as pd  df = pd.DataFrame({ 'Student': ['A','B','C','D','E','F','G','H'],  'Department': ['CSE','CSE','ECE','ECE','CSE','ECE','CSE','ECE'],  'Gender': ['F','M','F','M','M','F','F','M'],  'Marks': [85,72,91,65,78,88,95,70], 'Attendance': [92,81,96,72,85,90,98,75] })  

```python
import pandas as pd

df = pd.DataFrame({
    'Student': ['A','B','C','D','E','F','G','H'],
    'Department': ['CSE','CSE','ECE','ECE','CSE','ECE','CSE','ECE'],
    'Gender': ['M','F','F','M','F','F','M','F'],
    'Marks': [85,72,91,65,78,88,95,70],
    'Attendance': [92,81,96,72,85,90,98,75]
})
avg = df.groupby('Department')['Marks'].mean()
gender_avg = df.groupby(
    ['Department', 'Gender']
)['Marks'].mean()
highest = df.loc[
    df.groupby('Department')['Marks'].idxmax()
]
mean = df.groupby(
    'Department'
)['Marks'].transform('mean')

above_average = df[
    df['Marks'] > mean
]
attendance_avg = df.groupby(
    'Department'
)['Attendance'].mean()

highest_attendance = attendance_avg.idxmax()

print("1. Department Average:")
print(avg)

print("\n2. Department + Gender Average:")
print(gender_avg)

print("\n3. Highest Marks:")
print(highest)

print("\n4. Above Department Average:")
print(above_average)

print("\n5. Highest Average Attendance:")
print(attendance_avg)

print("\nDepartment:", highest_attendance)
```

Q4.  df = pd.DataFrame({ 'ID': [101,102,103,101,104,102,105],  'Name': ['A','B','C','A','D','B','E'], 'Age': [20,21,22,20,23,21,24],  'Score': [80,75,90,85,88,75,92] })  Perform: 1. Identify duplicated IDs. 2. For each ID, determine whether its records are identical. 3. If multiple records exist for an ID, retain the record having the highest Score. 4. Produce a clean DataFrame containing exactly one record per ID. 5. Report how many records were removed.

```python
import pandas as pd

df = pd.DataFrame({
    'ID': [101,102,103,101,104,102,105],
    'Name': ['A','B','C','D','E','F','G'],
    'Age': [20,21,22,23,21,24,21],
    'Score': [80,75,90,85,88,75,92]
})


duplicated_ids = df.loc[
    df['ID'].duplicated(keep=False), 'ID'
].unique()

print("Duplicated IDs:")
print(duplicated_ids)

for id_value in duplicated_ids:
    records = df[df['ID'] == id_value]

    identical = records.drop(
        columns='ID'
    ).nunique().eq(1).all()

    print("\nID:", id_value)
    print(records)

    if identical:
        print("Records are identical")
    else:
        print("Records are not identical")
clean_df = (
    df.sort_values(
        ['ID', 'Score'],
        ascending=[True, False],
        kind='stable'
    )
    .drop_duplicates('ID', keep='first')
    .sort_values('ID')
    .reset_index(drop=True)
)

print("\nClean DataFrame:")
print(clean_df)


removed = len(df) - len(clean_df)

print("\nRecords removed:", removed)
```

Q5.  df = pd.DataFrame({ 'Date': [ '2026-01-01', '2026-01-02', '2026-01-04', '2026-01-05', '2026-01-08', '2026-01-09' ],  'Sales': [100,120,150,130,180,200] })  Perform: 1. Convert Date into a proper datetime column. 2. Detect the missing dates. 3. Reindex the DataFrame to contain every date between the minimum and maximum date. 4. Keep missing sales as NaN.

```python
import pandas as pd

df = pd.DataFrame({
    'Date': [
        '2026-01-01',
        '2026-01-02',
        '2026-01-04',
        '2026-01-05',
        '2026-01-08',
        '2026-01-09'
    ],
    'Sales': [100,120,150,130,180,200]
})
df['Date'] = pd.to_datetime(df['Date'])

full_dates = pd.date_range(
    df['Date'].min(),
    df['Date'].max(),
    freq='D'
)

missing_dates = full_dates.difference(
    df['Date']
)

print("Missing dates:")
print(missing_dates)

df = df.set_index('Date')
df = df.reindex(full_dates)
df.index.name = 'Date'

df['Rolling_Average'] = (
    df['Sales']
    .rolling(window=3, min_periods=1)
    .mean()
)

print("\nDataFrame:")
print(df)

date = df['Rolling_Average'].idxmax()

print("\nMaximum rolling average date:")
print(date)

print("\nMaximum rolling average:")
print(df['Rolling_Average'].max())
```

Q6.  P = np.array([ [1, 2], [4, 6], [7, 3], [2, 8], [9, 5] ])  Construct a 5×5 matrix where element (i,j) contains the Euclidean distance between points i and j. The solution must: ● use NumPy broadcasting; ● contain no for or while loop; ● have zeros on the diagonal; ● produce a symmetric matrix. Then identify the pair of points having the maximum distance.

```python
import numpy as np

P = np.array([
    [1, 2],
    [4, 6],
    [7, 3],
    [2, 8],
    [9, 5]
])

diff = P[:, np.newaxis, :] - P[np.newaxis, :, :]

D = np.sqrt(
    np.sum(diff ** 2, axis=2)
)
np.fill_diagonal(D, 0)

print("Distance matrix:")
print(D)
i, j = np.unravel_index(
    np.argmax(D),
    D.shape
)

print("\nMaximum distance:")
print(D[i, j])

print("\nPair of points:")
print(i, j)

print("\nPoint", i, ":", P[i])
print("Point", j, ":", P[j])
```
