## Variable Declaration in Python

In Python, you don't need to declare a variable's type explicitly. You create a variable the moment you first assign a value to it using the assignment operator `=`. Python is dynamically typed, meaning the type of a variable is determined at runtime based on the value assigned to it.

```
# Declaring variables and assigning different data types

# Integer variable
my_integer = 10
print(f"my_integer: {my_integer}, Type: {type(my_integer)}")

# Float variable
my_float = 3.14
print(f"my_float: {my_float}, Type: {type(my_float)}")

# String variable
my_string = "Hello, Python!"
print(f"my_string: {my_string}, Type: {type(my_string)}")

# Boolean variable
is_true = True
print(f"is_true: {is_true}, Type: {type(is_true)}")

# List variable
my_list = [1, 2, 3, 'a', 'b']
print(f"my_list: {my_list}, Type: {type(my_list)}")

# Dictionary variable
my_dict = {'name': 'Alice', 'age': 30}
print(f"my_dict: {my_dict}, Type: {type(my_dict)}")

# Reassigning a variable (type can change)
my_variable = "Initial string value"
print(f"my_variable (initial): {my_variable}, Type: {type(my_variable)}")

my_variable = 123
print(f"my_variable (reassigned): {my_variable}, Type: {type(my_variable)}")
```

### Mutability in Python

In Python, objects can be either **mutable** or **immutable**:

*   **Mutable objects** can be changed after they are created. Examples include lists, dictionaries, and sets.
*   **Immutable objects** cannot be changed after they are created. Any operation that seems to change an immutable object actually creates a new object. Examples include numbers (integers, floats), strings, and tuples.

```
None
1
1.0
add
(1,'a',3,'b')
[1,'a',3,'b']
```

```
# Helper function to check mutability
def is_mutable(obj):
    if isinstance(obj, (list, dict, set)):
        return True
    # For other types, we can infer immutability unless proven otherwise
    # Strings, numbers, tuples, frozensets, NoneType, functions are generally immutable
    return False


items = [
    None,
    1,
    1.0,
    add, # Assuming 'add' function is defined in previous cells
    (1, 'a', 3, 'b'),
    [1, 'a', 3, 'b']
]

for item in items:
    item_type = type(item)
    mutability = "mutable" if is_mutable(item) else "immutable"
    print(f"Value: {item}, Type: {item_type}, Mutability: {mutability}")


# Special case for function, as 'add' can be ambiguous if multiple definitions exist.
# Let's explicitly check the last defined 'add' function if it exists.
if 'add' in globals() and callable(add):
    print(f"Value: add, Type: {type(add)}, Mutability: immutable (functions are immutable)")
else:
    print("Note: 'add' function definition was not found or is not callable in the current scope.")
```
